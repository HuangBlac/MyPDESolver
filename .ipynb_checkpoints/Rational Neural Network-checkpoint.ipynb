```python
import torch
import torch.nn as nn
import torch.optim as optim
from scipy.optimize import minimize

class ERA_PartialFractions(nn.Module):
    """基于部分分式分解的ERA激活函数"""
    def __init__(self, epsilon=1e-6):
        super(ERA_PartialFractions, self).__init__()
        self.epsilon = epsilon
        
        # 线性部分参数
        self.alpha = nn.Parameter(torch.randn(1))
        self.beta = nn.Parameter(torch.randn(1))
        
        # 部分分数参数
        self.p1 = nn.Parameter(torch.randn(1))
        self.q1 = nn.Parameter(torch.randn(1))
        self.p2 = nn.Parameter(torch.randn(1))
        self.q2 = nn.Parameter(torch.randn(1))
        
        # 分母参数 (确保初始时为正)
        self.c1 = nn.Parameter(torch.randn(1))
        self.d1 = nn.Parameter(torch.randn(1).abs() + 0.1)
        self.c2 = nn.Parameter(torch.randn(1))
        self.d2 = nn.Parameter(torch.randn(1).abs() + 0.1)
    
    def forward(self, x):
        # 分母多项式，确保不为零
        Q1 = (x - self.c1)**2 + self.d1**2 + self.epsilon
        Q2 = (x - self.c2)**2 + self.d2**2 + self.epsilon
        
        # 部分分数项
        frac1 = (self.p1 * x + self.q1) / Q1
        frac2 = (self.p2 * x + self.q2) / Q2
        
        # 完整ERA函数
        return self.alpha * x + self.beta + frac1 + frac2


class RationalNNWithERA(nn.Module):
    """结合ERA激活函数的有理神经网络"""
    def __init__(self, input_dim, hidden_dim):
        super().__init__()
        # 神经网络部分
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        
        # ERA激活函数
        self.era = ERA_PartialFractions()
        
        # 中间变量参数
        self.alpha = nn.Parameter(torch.randn(hidden_dim))
        self.beta = nn.Parameter(torch.randn(1))
        
        # 有理函数参数
        self.a = nn.Parameter(torch.randn(2))  # 线性分子
        self.b = nn.Parameter(torch.randn(2))  # 线性分母
        self.q1 = nn.Parameter(torch.randn(1))
        self.q2 = nn.Parameter(torch.ones(1))  # 初始为正，避免除零
        
        # 初始化策略：近似Swish函数
        self.initialize_as_swish()
    
    def initialize_as_swish(self):
        """将ERA初始化为近似Swish函数"""
        with torch.no_grad():
            # 近似Swish函数 x * sigmoid(x)
            self.era.alpha.fill_(0.5)   # 线性系数
            self.era.beta.fill_(0)      # 偏置
            
            # 部分分数参数设置
            self.era.p1.fill_(0.1)
            self.era.q1.fill_(0)
            self.era.p2.fill_(0)
            self.era.q2.fill_(0)
            
            # 分母参数设置为远离0的值，保证稳定性
            self.era.c1.fill_(0)
            self.era.d1.fill_(1)
            self.era.c2.fill_(0)
            self.era.d2.fill_(1)
            
            # 有理函数初始化为近似恒等映射
            self.a.data = torch.tensor([0.0, 1.0])
            self.b.data = torch.tensor([1.0, 0.0])
            self.q1.fill_(0.0)
            self.q2.fill_(1.0)
    
    def forward(self, x):
        # 前向传播：全连接 -> ERA激活 -> 中间变量 -> 有理函数输出
        h = self.era(self.fc1(x))
        z = torch.matmul(h, self.alpha) + self.beta  # 计算中间变量z
        
        # 有理函数部分
        p1 = self.a[0] + self.a[1] * z  # 线性分子
        p2 = self.b[0] + self.b[1] * z + 1e-6  # 线性分母，加小值避免除零
        
        return (p1 + self.q1) / (p2 + self.q2)
    
    def get_z(self, X):
        """获取隐藏层输出z，用于有理函数参数优化"""
        with torch.no_grad():
            h = self.era(self.fc1(X))
            z = torch.matmul(h, self.alpha) + self.beta
        return z
    
    def rational_params(self):
        """返回有理函数参数供SciPy优化"""
        return torch.cat([self.a.data, self.b.data, self.q1.data, self.q2.data]).numpy()
    
    def update_rational_params(self, new_params):
        """更新有理函数参数"""
        with torch.no_grad():
            self.a.data = torch.from_numpy(new_params[:2]).float()
            self.b.data = torch.from_numpy(new_params[2:4]).float()
            self.q1.data = torch.from_numpy(new_params[4:5]).float()
            self.q2.data = torch.from_numpy(new_params[5:6]).float()


def alt_train_with_era(model, X, y_true, max_iter=100, verbose=True):
    """交替训练模型：先优化神经网络参数，再优化有理函数参数"""
    # 确保输入是torch张量
    if not isinstance(X, torch.Tensor):
        X = torch.tensor(X, dtype=torch.float32)
    if not isinstance(y_true, torch.Tensor):
        y_true = torch.tensor(y_true, dtype=torch.float32)
    
    # 阶段一优化器 - 只优化神经网络参数(不包括有理函数参数)
    optimizer = optim.Adam([
        {'params': model.fc1.parameters()},
        {'params': model.alpha},
        {'params': model.beta},
        {'params': model.era.parameters()}  # 包含ERA参数
    ], lr=1e-3)
    
    # 记录损失
    losses = []
    
    for i in range(max_iter):
        # 阶段一：ADAM优化神经网络参数和ERA参数
        optimizer.zero_grad()
        y_pred = model(X)
        loss = torch.max(torch.abs(y_true - y_pred))  # 极小极大损失
        loss.backward()
        
        # 权重裁剪，稳定训练
        torch.nn.utils.clip_grad_value_(model.parameters(), clip_value=1.0)
        
        optimizer.step()
        
        # 阶段二：拟凸优化有理函数参数
        z = model.get_z(X).detach().numpy()
        
        # 定义优化目标函数：最小化最大误差
        def max_error(params):
            model.update_rational_params(params)
            with torch.no_grad():
                # 直接使用z计算，避免重复计算网络部分
                y_pred = (model.a[0] + model.a[1] * z + model.q1) / \
                         (model.b[0] + model.b[1] * z + model.q2 + 1e-6)
            return torch.max(torch.abs(torch.tensor(y_true.numpy() - y_pred.numpy()))).item()
        
        # 使用SLSQP优化器
        bounds = [(None, None)] * 6  # 所有参数无约束
        res = minimize(max_error, model.rational_params(), 
                      method='SLSQP', bounds=bounds, options={'maxiter': 50})
        
        # 更新有理函数参数
        model.update_rational_params(res.x)
        
        # 记录并打印损失
        losses.append(loss.item())
        if verbose and (i % 100 == 0):
            print(f"Iteration {i}, Loss: {loss.item():.6f}")
    
    return losses

```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from scipy.optimize import minimize
import matplotlib.pyplot as plt

class ERA_PartialFractions(nn.Module):
    """基于部分分式分解的ERA激活函数"""
    def __init__(self, epsilon=1e-6):
        super(ERA_PartialFractions, self).__init__()
        self.epsilon = epsilon
        
        # 线性部分参数
        self.alpha = nn.Parameter(torch.randn(1))
        self.beta = nn.Parameter(torch.randn(1))
        
        # 部分分数参数
        self.p1 = nn.Parameter(torch.randn(1))
        self.q1 = nn.Parameter(torch.randn(1))
        self.p2 = nn.Parameter(torch.randn(1))
        self.q2 = nn.Parameter(torch.randn(1))
        
        # 分母参数 (确保初始时为正)
        self.c1 = nn.Parameter(torch.randn(1))
        self.d1 = nn.Parameter(torch.randn(1).abs() + 0.1)
        self.c2 = nn.Parameter(torch.randn(1))
        self.d2 = nn.Parameter(torch.randn(1).abs() + 0.1)
    
    def forward(self, x):
        # 分母多项式，确保不为零
        Q1 = (x - self.c1)**2 + self.d1**2 + self.epsilon
        Q2 = (x - self.c2)**2 + self.d2**2 + self.epsilon
        
        # 部分分数项
        frac1 = (self.p1 * x + self.q1) / Q1
        frac2 = (self.p2 * x + self.q2) / Q2
        
        # 完整ERA函数
        return self.alpha * x + self.beta + frac1 + frac2


class RationalNNWithERA(nn.Module):
    """结合ERA激活函数的有理神经网络"""
    def __init__(self, input_dim, hidden_dim):
        super().__init__()
        # 神经网络部分
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        
        # ERA激活函数
        self.era = ERA_PartialFractions()
        
        # 中间变量参数
        self.alpha = nn.Parameter(torch.randn(hidden_dim))
        self.beta = nn.Parameter(torch.randn(1))
        
        # 有理函数参数
        self.a = nn.Parameter(torch.randn(2))  # 线性分子
        self.b = nn.Parameter(torch.randn(2))  # 线性分母
        self.q1 = nn.Parameter(torch.randn(1))
        self.q2 = nn.Parameter(torch.ones(1))  # 初始为正，避免除零
        
        # 初始化策略：近似Swish函数
        self.initialize_as_swish()
    
    def initialize_as_swish(self):
        """将ERA初始化为近似Swish函数"""
        with torch.no_grad():
            # 近似Swish函数 x * sigmoid(x)
            self.era.alpha.fill_(0.5)   # 线性系数
            self.era.beta.fill_(0)      # 偏置
            
            # 部分分数参数设置
            self.era.p1.fill_(0.1)
            self.era.q1.fill_(0)
            self.era.p2.fill_(0)
            self.era.q2.fill_(0)
            
            # 分母参数设置为远离0的值，保证稳定性
            self.era.c1.fill_(0)
            self.era.d1.fill_(1)
            self.era.c2.fill_(0)
            self.era.d2.fill_(1)
            
            # 有理函数初始化为近似恒等映射
            self.a.data = torch.tensor([0.0, 1.0])
            self.b.data = torch.tensor([1.0, 0.0])
            self.q1.fill_(0.0)
            self.q2.fill_(1.0)
    
    def forward(self, x):
        # 前向传播：全连接 -> ERA激活 -> 中间变量 -> 有理函数输出
        h = self.era(self.fc1(x))
        z = torch.matmul(h, self.alpha) + self.beta  # 计算中间变量z
        
        # 有理函数部分
        p1 = self.a[0] + self.a[1] * z  # 线性分子
        p2 = self.b[0] + self.b[1] * z + 1e-6  # 线性分母，加小值避免除零
        
        return (p1 + self.q1) / (p2 + self.q2)
    
    def get_z(self, X):
        """获取隐藏层输出z，用于有理函数参数优化"""
        with torch.no_grad():
            h = self.era(self.fc1(X))
            z = torch.matmul(h, self.alpha) + self.beta
        return z
    
    def rational_params(self):
        """返回有理函数参数供SciPy优化"""
        return torch.cat([self.a.data, self.b.data, self.q1.data, self.q2.data]).numpy()
    
    def update_rational_params(self, new_params):
        """更新有理函数参数"""
        with torch.no_grad():
            self.a.data = torch.from_numpy(new_params[:2]).float()
            self.b.data = torch.from_numpy(new_params[2:4]).float()
            self.q1.data = torch.from_numpy(new_params[4:5]).float()
            self.q2.data = torch.from_numpy(new_params[5:6]).float()


def alt_train_with_era(model, X, y_true, max_iter=100, verbose=True):
    """交替训练模型：先优化神经网络参数，再优化有理函数参数"""
    # 确保输入是torch张量
    if not isinstance(X, torch.Tensor):
        X = torch.tensor(X, dtype=torch.float32)
    if not isinstance(y_true, torch.Tensor):
        y_true = torch.tensor(y_true, dtype=torch.float32)
    
    # 阶段一优化器 - 只优化神经网络参数(不包括有理函数参数)
    optimizer = optim.Adam([
        {'params': model.fc1.parameters()},
        {'params': model.alpha},
        {'params': model.beta},
        {'params': model.era.parameters()}  # 包含ERA参数
    ], lr=1e-3)
    
    # 记录损失和结果
    losses = []
    results = {'training': {'loss_history': []}}
    
    for i in range(max_iter):
        # 阶段一：ADAM优化神经网络参数和ERA参数
        optimizer.zero_grad()
        y_pred = model(X)
        loss = torch.max(torch.abs(y_true - y_pred))  # 极小极大损失
        loss.backward()
        
        # 权重裁剪，稳定训练
        torch.nn.utils.clip_grad_value_(model.parameters(), clip_value=1.0)
        
        optimizer.step()
        
        # 阶段二：拟凸优化有理函数参数
        z = model.get_z(X).detach().numpy()
        
        # 定义优化目标函数：最小化最大误差
        def max_error(params):
            model.update_rational_params(params)
            with torch.no_grad():
                # 直接使用z计算，避免重复计算网络部分
                y_pred = (model.a[0] + model.a[1] * z + model.q1) / \
                         (model.b[0] + model.b[1] * z + model.q2 + 1e-6)
            return torch.max(torch.abs(torch.tensor(y_true.numpy() - y_pred.numpy()))).item()
        
        # 使用SLSQP优化器
        bounds = [(None, None)] * 6  # 所有参数无约束
        res = minimize(max_error, model.rational_params(), 
                      method='SLSQP', bounds=bounds, options={'maxiter': 50})
        
        # 更新有理函数参数
        model.update_rational_params(res.x)
        
        # 记录并打印损失
        current_loss = loss.item()
        losses.append(current_loss)
        results['training']['loss_history'].append(current_loss)
        
        if verbose and (i % 100 == 0):
            print(f"Iteration {i}, Loss: {current_loss:.6f}")
    
    return losses, results


if __name__ == "__main__":
    # 创建示例数据 
    X = torch.linspace(-3, 3, 100).reshape(-1, 1)  # 1D输入
    y_true = torch.sqrt(torch.abs(X - 0.25)).flatten() 
    
    # 创建模型
    model = RationalNNWithERA(input_dim=1, hidden_dim=10)
    
    # 训练模型
    print("开始训练模型...")
    losses, results = alt_train_with_era(model, X, y_true, max_iter=5000)
    
    # 测试模型
    with torch.no_grad():
        y_pred = model(X)
        final_error = torch.mean(torch.abs(y_pred - y_true)) 
        max_error = torch.max(torch.abs(y_pred - y_true))
        print(f"最终平均绝对误差: {final_error.item():.6f}")
        print(f"最终最大绝对误差: {max_error.item():.6f}")
    
    # 绘制训练损失对比图
    plt.figure(figsize=(10, 6))
    for name, metrics in results.items():
        plt.plot(metrics['loss_history'], label=name, alpha=0.7)
    plt.xlabel('Epoch')
    plt.ylabel('L1 Loss')
    plt.title('Training Loss Comparison')
    plt.legend()
    plt.grid(True)
    plt.show()
    
    # 添加预测结果与真实值对比图
    plt.figure(figsize=(10, 6))
    plt.scatter(X.numpy(), y_true.numpy(), label='真实值', alpha=0.5)
    plt.plot(X.numpy(), y_pred.numpy(), 'r-', label='预测值', linewidth=2)
    plt.xlabel('X')
    plt.ylabel('y')
    plt.title('预测结果与真实值对比')
    plt.legend()
    plt.grid(True)
    plt.show()
    
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from tqdm import tqdm

# 基线模型1: 标准全连接神经网络 (ReLU激活)
class FCNet_ReLU(nn.Module):
    def __init__(self, input_dim, hidden_dim):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, 1)  # 回归任务输出1维
        )
        
    def forward(self, x):
        return self.model(x).flatten()  # 展平输出


# 基线模型2: 全连接网络 (Swish激活)
class FCNet_Swish(nn.Module):
    def __init__(self, input_dim, hidden_dim):
        super().__init__()
        self.hidden1 = nn.Linear(input_dim, hidden_dim)
        self.hidden2 = nn.Linear(hidden_dim, hidden_dim)
        self.output = nn.Linear(hidden_dim, 1)
        
    def forward(self, x):
        x = self.hidden1(x)
        x = x * torch.sigmoid(x)  # Swish激活
        x = self.hidden2(x)
        x = x * torch.sigmoid(x)  # Swish激活
        return self.output(x).flatten()


# 基线模型3: 简化版有理网络 (仅使用基础激活函数)
class SimpleRationalNet(nn.Module):
    def __init__(self, input_dim, hidden_dim):
        super().__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.alpha = nn.Parameter(torch.randn(hidden_dim))
        self.beta = nn.Parameter(torch.randn(1))
        
        # 简单有理函数输出层
        self.a = nn.Parameter(torch.tensor([0.0, 1.0]))  # 分子系数
        self.b = nn.Parameter(torch.tensor([1.0, 0.0]))  # 分母系数
        
    def forward(self, x):
        h = self.relu(self.fc1(x))
        z = torch.matmul(h, self.alpha) + self.beta
        return (self.a[0] + self.a[1] * z) / (self.b[0] + self.b[1] * z + 1e-6)


# 统一训练函数 (适用于所有基线模型)
def train_baseline(model, X, y_true, epochs=5000, lr=1e-3, verbose=True):
    if not isinstance(X, torch.Tensor):
        X = torch.tensor(X, dtype=torch.float32)
    if not isinstance(y_true, torch.Tensor):
        y_true = torch.tensor(y_true, dtype=torch.float32)
    
    optimizer = optim.Adam(model.parameters(), lr=lr)
    criterion = nn.L1Loss()  # 使用L1损失与原模型的MAE对比
    losses = []
    
    for epoch in tqdm(range(epochs), disable=not verbose):
        optimizer.zero_grad()
        y_pred = model(X)
        loss = criterion(y_pred, y_true)
        loss.backward()
        optimizer.step()
        
        losses.append(loss.item())
        if verbose and (epoch % 50 == 0):
            print(f"Epoch {epoch}, Loss: {loss.item():.6f}")
    
    return losses


# 模型比较实验
if __name__ == "__main__":
    # 生成与之前相同的数据集
    X = torch.linspace(-3, 3, 100).reshape(-1, 1)
    y_true = torch.sqrt(abs(X-0.25)).flatten()  + 0.1 * torch.randn(100)
    
    # 初始化所有模型 (保持相同的隐藏层维度)
    hidden_dim = 10
    models = {
        "FC-ReLU": FCNet_ReLU(1, hidden_dim),
        "FC-Swish": FCNet_Swish(1, hidden_dim),
        "Simple-Rational": SimpleRationalNet(1, hidden_dim)
    }
    
    # 训练并评估所有模型
    results = {}
    for name, model in models.items():
        print(f"\n训练 {name}...")
        losses = train_baseline(model, X, y_true, epochs=5000)
        
        # 评估最终性能
        with torch.no_grad():
            y_pred = model(X)
            mae = torch.mean(torch.abs(y_pred - y_true)).item()
            max_error = torch.max(torch.abs(y_pred - y_true)).item()
        
        results[name] = {
            "final_mae": mae,
            "max_error": max_error,
            "last_loss": losses[-1]
        }
        print(f"{name} 结果: MAE={mae:.6f}, 最大误差={max_error:.6f}")
    
    # 打印比较结果
    print("\n===== 模型比较结果 =====")
    for name, metrics in results.items():
        print(f"{name}:")
        print(f"  平均绝对误差: {metrics['final_mae']:.6f}")
        print(f"  最大误差: {metrics['max_error']:.6f}")
    
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
import seaborn as sns
from rational_nn_with_era import RationalNNWithERA, alt_train_with_era
from baseline_models import FCNet_ReLU, FCNet_Swish, SimpleRationalNet, train_baseline

# 设置中文字体，确保中文正常显示
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False  # 正确显示负号

def generate_data(n_samples=100, noise_level=0.1):
    """生成用于测试的正弦曲线数据"""
    X = torch.linspace(-3, 3, n_samples).reshape(-1, 1)
    y_true = torch.sin(X).flatten() + noise_level * torch.randn(n_samples)
    return X, y_true

def train_all_models(X, y_true, hidden_dim=10):
    """训练所有模型并返回损失记录"""
    # 初始化模型
    models = {
        "有理网络(ERA)": RationalNNWithERA(input_dim=1, hidden_dim=hidden_dim),
        "全连接(ReLU)": FCNet_ReLU(1, hidden_dim),
        "全连接(Swish)": FCNet_Swish(1, hidden_dim),
        "简单有理网络": SimpleRationalNet(1, hidden_dim)
    }
    
    # 训练模型并记录损失
    loss_records = {}
    
    # 训练有理网络(ERA)
    print("训练有理网络(ERA)...")
    loss_records["有理网络(ERA)"] = alt_train_with_era(
        models["有理网络(ERA)"], X, y_true, max_iter=100, verbose=False)
    
    # 训练基线模型
    for name in ["全连接(ReLU)", "全连接(Swish)", "简单有理网络"]:
        print(f"训练{name}...")
        loss_records[name] = train_baseline(
            models[name], X, y_true, epochs=500, verbose=False)
    
    return models, loss_records

def plot_loss_curves(loss_records, title="不同模型的损失曲线对比"):
    """绘制所有模型的损失曲线"""
    plt.figure(figsize=(10, 6))
    sns.set_style("whitegrid")
    
    # 定义不同的线条样式和颜色
    styles = {
        "有理网络(ERA)": {"color": "#2E91E5", "linestyle": "-", "linewidth": 2},
        "全连接(ReLU)": {"color": "#E01E5A", "linestyle": "--", "linewidth": 2},
        "全连接(Swish)": {"color": "#1FE082", "linestyle": "-.", "linewidth": 2},
        "简单有理网络": {"color": "#792387", "linestyle": ":", "linewidth": 2}
    }
    
    # 绘制每条损失曲线
    for name, losses in loss_records.items():
        # 为了更好的可视化，对长曲线进行下采样
        if len(losses) > 200:
            step = len(losses) // 200
            losses = losses[::step]
            x = np.arange(0, len(losses)*step, step)
        else:
            x = np.arange(len(losses))
            
        plt.plot(x, losses, label=name, **styles[name])
    
    # 设置图表属性
    plt.title(title, fontsize=14)
    plt.xlabel("训练迭代次数", fontsize=12)
    plt.ylabel("损失值 (L1 Loss)", fontsize=12)
    plt.yscale("log")  # 使用对数刻度更好地展示损失变化
    plt.legend(fontsize=11)
    plt.xlim(0, max(len(losses) for losses in loss_records.values()))
    plt.gca().xaxis.set_major_locator(MaxNLocator(integer=True))  # 确保x轴为整数
    plt.tight_layout()
    
    return plt.gcf()

def plot_predictions(models, X, y_true, title="不同模型的预测结果对比"):
    """绘制所有模型的预测结果与真实值对比"""
    plt.figure(figsize=(10, 6))
    sns.set_style("whitegrid")
    
    # 绘制真实值
    plt.scatter(X.numpy(), y_true.numpy(), label="真实值", color="#6B7280", alpha=0.5, s=30)
    
    # 绘制各模型预测值
    colors = {
        "有理网络(ERA)": "#2E91E5",
        "全连接(ReLU)": "#E01E5A",
        "全连接(Swish)": "#1FE082",
        "简单有理网络": "#792387"
    }
    
    with torch.no_grad():
        for name, model in models.items():
            y_pred = model(X)
            plt.plot(X.numpy(), y_pred.numpy(), label=name, color=colors[name], linewidth=2.5)
    
    # 设置图表属性
    plt.title(title, fontsize=14)
    plt.xlabel("X值", fontsize=12)
    plt.ylabel("Y值", fontsize=12)
    plt.legend(fontsize=11)
    plt.tight_layout()
    
    return plt.gcf()

if __name__ == "__main__":
    # 生成数据
    X, y_true = generate_data(n_samples=100, noise_level=0.1)
    
    # 训练所有模型
    models, loss_records = train_all_models(X, y_true)
    
    # 绘制损失曲线并保存
    loss_fig = plot_loss_curves(loss_records)
    loss_fig.savefig("模型损失曲线对比.png", dpi=300, bbox_inches="tight")
    
    # 绘制预测结果并保存
    pred_fig = plot_predictions(models, X, y_true)
    pred_fig.savefig("模型预测结果对比.png", dpi=300, bbox_inches="tight")
    
    print("图表已保存为PNG文件")
    plt.show()
    
```
